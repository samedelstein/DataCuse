```python
import pandas as pd
```

```python
from pathlib import Path

data_name = "Syracuse_Tree_Data_7832308158023752279.csv"
candidates = [Path.cwd(), Path.cwd() / "analysis", Path.cwd() / "1_oak_street_trees" / "analysis", Path.cwd() / "content" / "1_oak_street_trees" / "analysis"]
analysis_dir = next((folder for folder in candidates if (folder / data_name).is_file()), None)
if analysis_dir is None:
    raise FileNotFoundError("Run from the DataCuse repo, content folder, story folder, or analysis folder.")
trees = pd.read_csv(analysis_dir / data_name)
```

```python
oak_street_trees = trees[
  trees["SPP_COM"].str.contains("oak", case=False, na=False)
  & trees["STREET"].str.contains("oak", case=False, na=False)
]
```

```python
oak_street_trees
```

```python
tree_names = (
      trees["SPP_COM"]
      .fillna("")
      .str.split(",").str[0]
      .str.strip()
      .str.lower()
  )

street_names = trees["STREET"].fillna("").str.lower()

mask = [
  bool(tree) and tree in street
  for tree, street in zip(tree_names, street_names)
]

matches = trees[mask]
print(matches[["SPP_COM", "STREET"]])
```

```python
import re

mask = [
  bool(tree) and bool(re.search(r"\b" + re.escape(tree) + r"\b", street))
  for tree, street in zip(tree_names, street_names)
]

matches = trees[mask]
matches
```
